```python

import pandas as pd

# Nama lengkap wilayah dari data konsumsi
demand_regions = set(
    df_konsumsi_wilayah["region"].astype(str).str.strip()
)

# Daftar nama wilayah pada data produksi
production_regions = sorted(
    df_produksi["region"].astype(str).str.strip().unique()
)

mapping_rows = []
unmatched_regions = []

for production_region in production_regions:

    # Kota pada data produksi sudah memiliki prefix.
    if production_region.startswith("Kota "):
        demand_region = production_region

    # Nama pendek dicocokkan dengan nama kabupaten.
    elif f"Kabupaten {production_region}" in demand_regions:
        demand_region = f"Kabupaten {production_region}"

    # Fallback jika nama ternyata sudah sama.
    elif production_region in demand_regions:
        demand_region = production_region

    else:
        unmatched_regions.append(production_region)
        continue

    mapping_rows.append({
        "region_id": (
            demand_region.upper()
            .replace("KABUPATEN ", "KAB_")
            .replace("KOTA ", "KOTA_")
            .replace(" ", "_")
    ),
        "region_standard": demand_region,
        "production_region": production_region,
        "demand_region": demand_region
    })

region_master = pd.DataFrame(mapping_rows)

# Validasi mapping
assert not unmatched_regions, (
    f"Wilayah produksi belum cocok: {unmatched_regions}"
)

assert region_master["region_id"].is_unique, (
    "region_id tidak unik."
)

assert region_master["demand_region"].is_unique, (
    "Ada wilayah demand yang dipetakan lebih dari sekali."
)

assert len(region_master) == 38, (
    f"Jumlah wilayah hasil mapping: {len(region_master)}, bukan 38."
)

region_master = region_master.sort_values(
    "region_standard"
).reset_index(drop=True)

print("Region master berhasil dibuat.")
print("Jumlah wilayah:", len(region_master))
display(region_master)
```

```python

# Buat pemetaan nama wilayah produksi -> konsumsi
wilayah_konsumsi = set(
    df_konsumsi_wilayah["region"].astype(str).str.strip()
)

def cocokkan_wilayah(nama):
    nama = str(nama).strip()

    # Jika nama sudah cocok persis, gunakan langsung.
    if nama in wilayah_konsumsi:
        return nama

    # Nama produksi pendek biasanya merujuk ke kabupaten.
    kandidat = f"Kabupaten {nama}"
    if kandidat in wilayah_konsumsi:
        return kandidat

    return None


region_mapping = (
    supply_result[["region"]]
    .drop_duplicates()
    .assign(
        region_demand=lambda x: x["region"].map(cocokkan_wilayah)
    )
)

print("Jumlah wilayah produksi:", len(region_mapping))
print("Wilayah belum cocok:")
display(region_mapping[region_mapping["region_demand"].isna()])
```

```python

# Terapkan pemetaan pada data produksi
supply_mapped = supply_result.merge(
    region_mapping,
    on="region",
    how="left",
    validate="many_to_one",
)

# Ubah nama wilayah produksi menjadi nama standar konsumsi.
supply_mapped = supply_mapped.drop(columns=["region"])
supply_mapped = supply_mapped.rename(
    columns={
        "region_demand": "region",
        "konsumsi": "konsumsi",
    }
)

# Siapkan baseline konsumsi dengan nama wilayah standar.
demand_baseline = df_konsumsi_wilayah.copy()
demand_baseline["region"] = (
    demand_baseline["region"].astype(str).str.strip()
)

# Gabungkan produksi dengan baseline konsumsi.
coordinator_result = supply_mapped.merge(
    demand_baseline,
    on="region",
    how="left",
    validate="many_to_one",
    suffixes=("", "_demand"),
)

print("Jumlah baris:", len(coordinator_result))
print("Jumlah wilayah:", coordinator_result["region"].nunique())
print("Tahun:", sorted(coordinator_result["year"].unique()))

display(
    coordinator_result[
        ["year", "region", "produksi", "konsumsi"]
    ].head(10)
)
```

```python

import pandas as pd

# 1. Siapkan baseline konsumsi
demand_baseline = df_konsumsi_wilayah.copy()
demand_baseline["region"] = (
    demand_baseline["region"].astype(str).str.strip()
)

# 2. Buat pemetaan nama produksi ke nama standar konsumsi
nama_konsumsi = set(demand_baseline["region"])

def map_region(nama):
    nama = str(nama).strip()

    # Prioritaskan kecocokan langsung, termasuk nama Kota.
    if nama in nama_konsumsi:
        return nama

    # Nama pendek yang belum cocok dipetakan ke Kabupaten.
    kandidat = f"Kabupaten {nama}"
    if kandidat in nama_konsumsi:
        return kandidat

    return None

region_mapping = (
    supply_result[["region"]]
    .drop_duplicates()
    .copy()
)

region_mapping["region_standard"] = (
    region_mapping["region"].map(map_region)
)

# 3. Hentikan proses jika ada nama yang belum cocok.
belum_cocok = region_mapping[
    region_mapping["region_standard"].isna()
]

if not belum_cocok.empty:
    raise ValueError(
        "Pemetaan wilayah belum lengkap:\n"
        + str(belum_cocok["region"].tolist())
    )

if region_mapping["region_standard"].nunique() != 38:
    raise ValueError(
        "Pemetaan tidak menghasilkan 38 wilayah yang unik. "
        "Periksa nama kabupaten dan kota."
    )

# 4. Terapkan nama standar ke hasil Supply Agent.
supply_mapped = supply_result.merge(
    region_mapping,
    on="region",
    how="left",
    validate="many_to_one",
)

supply_mapped = supply_mapped.drop(columns=["region"])
supply_mapped = supply_mapped.rename(
    columns={"region_standard": "region"}
)

# 5. Gabungkan dengan baseline konsumsi.
coordinator_result = supply_mapped.merge(
    demand_baseline,
    on="region",
    how="left",
    validate="many_to_one",
)

# 6. Validasi akhir.
assert len(coordinator_result) == len(supply_result)
assert coordinator_result["region"].nunique() == 38
assert coordinator_result["konsumsi"].notna().all()

print("Coordinator berhasil dibangun.")
print("Jumlah baris:", len(coordinator_result))
print("Jumlah wilayah:", coordinator_result["region"].nunique())
print("Periode:", sorted(coordinator_result["year"].unique()))

display(
    coordinator_result[
        ["year", "region", "produksi", "konsumsi", "supply_status"]
    ].head(10)
)
```
